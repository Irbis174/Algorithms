```python
import sys


def main():
    hash_table = {}
    
    st1 = input()
    st2 = input()

    for i in st1:
        hash_table[i] = hash_table.get(i, 0) + 1
    
    for i in st2:
        if i not in hash_table:
            print(0)
            return
        hash_table[i] -= 1
        if hash_table[i] == 0:
            del hash_table[i]

    print(int(hash_table != 0))


if __name__ == '__main__':
    main()
```
