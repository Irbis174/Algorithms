```python
class Solution:
    def maxArea(self, height: List[int]) -> int:
        left, right = 0, len(height) - 1
        mx = 0

        while left < right:
            mx = max(min(height[right], height[left]) * (right - left), mx)

            if height[right] > height[left]:
                left += 1
            else:
                right -= 1
            
        return mx
```
