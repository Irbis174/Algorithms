```python
class Solution:
    def minSubArrayLen(self, target: int, nums: List[int]) -> int:
        slow, fast = 0, 0
        mn = len(nums) + 1
        sm = 0

        for fast in range(len(nums)):
            sm += nums[fast]

            while sm >= target:
                
                mn = min(fast-slow+1, mn)
                sm -= nums[slow]  
                slow += 1
                

        return 0 if mn == len(nums) + 1 else mn
```
