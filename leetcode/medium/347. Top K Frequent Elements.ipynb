```python
class Solution:
    def topKFrequent(self, nums: List[int], k: int) -> List[int]:
        hash_table = {}
        res = []
        for i in nums:
            hash_table[i] = hash_table.get(i,0) + 1

        hash_table = sorted(hash_table, key=hash_table.get, reverse=True)

        for i, elem in enumerate(hash_table):
            res.append(elem)
            if i == k-1:
                return res
```
