```python
class Solution:
    def longestOnes(self, nums: List[int], k: int) -> int:
        slow, fast = 0,0 
        mx = 0
        zeros = 0

        for fast in range(len(nums)):
            if nums[fast] == 0:
                zeros += 1
            
            while zeros > k:
                if nums[slow] == 0:
                    zeros -= 1
                slow += 1

            mx = max(fast - slow + 1, mx)
        
        return mx
```
