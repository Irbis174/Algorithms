```python
#A. Выполнение операций со словарем

def main():
    n = int(input())
    hash_table = {}
    for i in range(n):
        tpl = tuple(map(int, input().split()))
        if tpl[0] == 1:
            hash_table[tpl[1]] = tpl[2]
        else:
            if tpl[1] in hash_table:
                print(hash_table[tpl[1]])
            else:
                print(-1)

main()
```

```python
#B. Поиск самого частого элемента в массиве

def main():
    n = int(input())
    arr = list(map(int, input().split()))
    hash_table = {}

    for i in arr:
        hash_table[i] = hash_table.get(i, 0) + 1

    mx = 0
    number = max(hash_table)
    for i in hash_table:
        if hash_table[i] > mx:
            mx = max(hash_table[i], mx)
            number = i
        elif hash_table[i] == mx:
            if i < number:
                number = i
    print(number)

main()
```
