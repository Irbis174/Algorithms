```python
# A. Выполнение операций с множеством

def main():
    n = int(input())
    mn = set()
    for i in range(n):
        a, b = tuple(map(int, input().split()))
        if a == 1:
            mn.add(b)
        else:
            if b in mn:
                print(1)
            else:
                print(0)

main()
```

```python
# B. Объединение множеств

def main():
    n = int(input())
    mn = set()
    for i in range(n):
        arr = list(map(int, input().split()))
        for j in range(1,len(arr)):
            mn.add(arr[j])
    print(len(mn))
    
main()
```

```python
# C. Пересечение множеств

def main():
    n = int(input())
    mn_arr = [set() for i in range(n)]
    for i in range(n):
        arr = list(map(int, input().split()))
        mn_arr[i] = set(arr[1:])
    print(len(set.intersection(*mn_arr)))
    
main()
```

```python
# D. Подсолнечник

def main():
    n = int(input())
    mn_arr = [set() for i in range(n)]
    for i in range(n):
        arr = list(map(int, input().split()))
        mn_arr[i] = set(arr[1:])

    if (len(set.intersection(*mn_arr)) == 0):
        print('NO')
    else:
        print('YES')
        mn = set.intersection(*mn_arr)
        print(len(mn)) 
        for i in range(len(mn_arr)):
            print(len(mn_arr[i] - mn), end = ' ') 

main()
```
