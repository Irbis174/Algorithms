```python
n = int(input())
hash_dict = {}

for i in range(n):
    name = input()
    if name not in hash_dict:
        print('OK')
        hash_dict[name] = []
        hash_dict[name].append(name) 
    else:
        new_name = name + str(len(hash_dict[name]))
        hash_dict[name].append(new_name)
        print(new_name)
```
