```python
n, t = tuple(map(int, input().split()))
arr = list(map(int, input().split()))

sm = 0
cnt = 0
slow, fast = 0, 0
mx = 0

for fast in range(n):
    
    sm += arr[fast]

    while sm > t:  
        sm -= arr[slow]
        slow += 1
    
    mx = max(mx, fast - slow + 1)


print(mx)

```
